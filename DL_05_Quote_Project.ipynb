## Preprocess the data

```python
quotes = quotes.str.lower()
```

```python
import string
translator = str.maketrans('', '', string.punctuation)

quotes = quotes.apply(lambda x: x.translate(translator))
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocab_size = 10000

tokenizer = Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index = tokenizer.word_index
print(len(word_index))
```

```python
list(tokenizer.word_index.items())[:10]
```

```python
sequences = tokenizer.texts_to_sequences(quotes)
```

```python
sequences
```

```python
X = []
y = []

for seq in sequences:
    for i in range(1, len(seq)):
        X.append(seq[:i])
        y.append(seq[i])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, Dense
```

```python
embedding_dim = 50
rnn_units = 128
```

```python
rnn_model = Sequential()
rnn_model.add(
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_len)
)
rnn_model.add(SimpleRNN(units=rnn_units, activation='tanh'))
rnn_model.add(Dense(
    units=vocab_size, activation='softmax'
))
```

```python
rnn_model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
rnn_model.summary()
```

```python
LSTM_model = Sequential()
LSTM_model.add(
    Embedding(input_dim=vocab_size, output_dim=embedding_dim, input_length=max_len)
)
LSTM_model.add(LSTM(units=rnn_units, activation='tanh'))
LSTM_model.add(Dense(
    units=vocab_size, activation='softmax'
))
LSTM_model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
LSTM_model.summary()
```

```python
from tensorflow.keras.models import load_model

lstm_model = load_model("lstm_model2.h5")
```

```python
index_to_word = {index: word for word, index in tokenizer.word_index.items()}
index_to_word  # Example: Get the word corresponding to index 1
```

```python
def predictor(model,tokenizer,text,max_len):
    text = text.lower()
    seq  = tokenizer.texts_to_sequences([text])[0]
    seq = pad_sequences([seq], maxlen=max_len, padding='pre')

    pred = model.predict(seq,verbose=0)
    pred_index = np.argmax(pred)
    return index_to_word.get(pred_index, "")  
```

```python
seed_text = "Life is a"
next_word = predictor(lstm_model, tokenizer, seed_text, max_len)
print(next_word)
```

```python
def generate_text(model, tokenizer, seed_text, max_len, num_words):
    generated_text = seed_text
    for _ in range(num_words):
        next_word = predictor(model, tokenizer, generated_text, max_len)
        if next_word == "":
            break  # Stop if no next word is predicted
        generated_text += " " + next_word
    return generated_text
```

```python
text = "Your discipline is your"
generated_text = generate_text(lstm_model, tokenizer, text, max_len, num_words=10)
print(generated_text)
```
